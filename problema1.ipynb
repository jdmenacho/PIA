```python
def formatearLista(listaEnteros):
    listaNueva = []
    for i in listaEnteros:
        if i > 0 and i not in listaNueva:
            listaNueva.append(i)

    listaNueva.sort()
    return listaNueva
    
#Hacemos un ejemplo
lista = [4, -1, 2, 4, 3, -5, 2]
#visualizamos lista de enteros
print(lista)

listaNueva = formatearLista(lista)
#visualizamos lista formateada
print(listaNueva)
            
        
```
