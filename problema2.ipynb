```python
import string

def crear_fichero(ruta):
    with open(ruta, "w", encoding="utf-8") as fichero:
        fichero.write("esto es un fichero de ejemplo con texto dentro\n")
        fichero.write("siguiendo el ejemplo, dentro del TEXTO, tengo que localizar PALABRAS \n")
        fichero.write("Espero que con el ejemplo pueda encontrar las palabras\n")
    print("Fichero creado correctamente")

def leer_fichero(ruta, lista_palabras):
    # Inicializamos el contador para cada palabra buscada
    contador = {palabra: 0 for palabra in lista_palabras}
    
    with open(ruta, "r", encoding="utf-8") as fichero:
        for linea in fichero:
            # Eliminar puntuación y convertir a minúsculas
            linea_limpia = linea.translate(str.maketrans('', '', string.punctuation)).lower()
            palabras_en_linea = linea_limpia.split()
            
            # Contar frecuencia de cada palabra buscada
            for palabra in lista_palabras:
                contador[palabra] += palabras_en_linea.count(palabra)
    
    
    return dict(sorted(contador.items()))

# Ruta del fichero
ruta = "fichero.txt"

# Creamos el fichero
crear_fichero(ruta)

# Contamos la frecuencia de palabras
lista_palabras = ["ejemplo", "palabra", "fichero", "texto"]
resultado = leer_fichero(ruta, lista_palabras)  
print(resultado)              
                
                
```
